```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x ** 2 - 3 * x + x * np.log(x)
```

```python
def дихотомия(a, b, epsilon):
    global a0, b0
    print(f'Дихотомия')
    a0, b0 = a, b
    delta = epsilon / 2
    counter = 0 
    while (b0 - a0) >= epsilon:
        x1, x2 = (a0 + b0 - delta) / 2, (a0 + b0 + delta) / 2
        f1, f2 = f(x1), f(x2)
        counter += 2

        print(f'[{a0}; {b0}]   {b0 - a0}')

        if f1 < f2:
            b0 = x2
        elif f1 > f2:
            a0 = x1
        else:
            a0, b0 = x1, x2

    print(f'[{a0}; {b0}]   {b0 - a0}')

    return (a0 + b0) / 2, f((a0 + b0) / 2), counter
```

```python
def золотое_сечение(a, b, epsilon):
    global a0, b0
    print('Золотое сечение')
    a0, b0 = a, b
    counter = 2 
    while (b0 - a0) >= epsilon:
        x1 = a0 + (b0 - a0) * (3 - np.sqrt(5)) / 2
        x2 = a0 + (b0 - a0) * (np.sqrt(5) - 1) / 2
        f1, f2 = f(x1), f(x2)
        counter += 1

        print(f'[{a0}; {b0}]   {b0 - a0}')

        if f1 < f2:
            b0 = x2
        elif f1 > f2:
            a0 = x1
        else:
            a0, b0 = x1, x2

    print(f'[{a0}; {b0}]   {b0 - a0}')

    return (a0 + b0) / 2, f((a0 + b0) / 2), counter
```

```python
a, b = 1, 2
epsilon = 0.001
```

```python
print(дихотомия(a, b, epsilon))
print(золотое_сечение(a, b, epsilon))
```

```python
x, y1, y2 = [], [], []
```

```python
for eps in np.arange(0.0001, 0.1, 0.000001):
    x.append(np.log(eps))
    y1.append(дихотомия(a, b, eps)[2])
    y2.append(золотое_сечение(a, b, eps)[2])
```

```python
plt.figure(figsize=(10, 8))

plt.plot(x, y1, label="Дихотомия")
plt.plot(x, y2, label="Золотое сечение")

plt.xlabel('$ln(\epsilon)$')
plt.ylabel('Число вызовов функции')

plt.legend()

plt.savefig('graph.png', dpi=300)
plt.show()
```
